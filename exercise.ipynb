```python
import numpy as np

#Columns: [age, math marks, science marks]
data = np.array([
    [18, 85, 78],  # student 1
    [19, 97, 88],  # student 2
    [17, 69, 90],  # student 3
    [20, 88, 85],  # student 4
    [16, 75, 80]   # student 5
])
```

```python
# get the shape of the matrix.
data.shape
```

```python
# find the average age of the students.
np.mean(data[:,0])
```

```python
# extract the math marks of all students.
data[:,1]
```

```python
# find the highest science marks among the students.
np.max(data[:,2])
```

```python
# get details of students who scored more than 90 in math.
data[data[:,1] > 90]
```

```python
# increase the math marks of all students by 5.
data[:,1] + 5
```

```python
# find how many students are younger than 19.
np.sum(data[:,0] < 19)
```

```python
# calculate the average math marks of students (column-wise mean).
np.mean(data[:,1:],axis =0)
```

```python
# get data of students who scored at least 80 in both the subjects.
data[(data[:,1] >= 80) & (data[:,2] >=80)]
```

```python
# replace all science marks less than < 75 with 0.
data[:,2][data[:,2] < 75] = 0
data
```
